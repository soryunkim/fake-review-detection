# **모델 성능 평균 시각화**

```python
# 각 모델 결과를 따로 저장한 뒤, 한 번에 시각화하기 위함
# 각 결과는 각 ipynb 파일 내 존재함
machine = pd.read_csv('machine_learning_result_mean_last.csv')
deep = pd.read_csv('deep_learning_mean_result.csv')
```

```python
machine
```

```python
deep
```

```python
result = pd.concat([machine[:3], deep], axis=0)
```

```python
result
```

```python
result.sort_values(by='mean', ascending=False).reset_index(drop=True)
```

```python
result.to_csv('result_last.csv', index=False)
```

```python
result = result.T
result.columns = result.iloc[0]
result = result.iloc[1:]
result = result[:4]
result
```

```python
# 그래프 크기 설정
plt.figure(figsize=(12, 8))

# Model 선택
last_df = result[['Gradient Boosting', 'Random Forest', 'XGBoost', 'CNN_LSTM', 'CNN', 'CNN_RNN']]

# X축의 위치 설정
x = np.arange(len(last_df))

# 막대그래프 그리기
bar_width = 0.1  # 막대의 너비

plt.bar(x, last_df['Gradient Boosting'], width=bar_width, label='Gradient Boosting', color='green')
plt.bar(x + bar_width, last_df['Random Forest'], width=bar_width, label='Random Forest', color='limegreen')
plt.bar(x + bar_width * 2, last_df['XGBoost'], width=bar_width, label='XGBoost', color='lime')


plt.bar(x + bar_width * 3, last_df['CNN_LSTM'], width=bar_width, label='CNN_LSTM', color='orange')
plt.bar(x + bar_width * 4, last_df['CNN'], width=bar_width, label='CNN', color='gold')
plt.bar(x + bar_width * 5, last_df['CNN_RNN'], width=bar_width, label='CNN_RNN', color='yellow')


plt.xticks(x + 0.3, ['Accuracy', 'Precision', 'Recall', 'F1-Score'])

# X축 레이블 설정
plt.xlabel('Metrics')
plt.ylabel('Scores')
plt.title('Comparison of Model Perfomance')

# 범례 표시
plt.legend(loc='lower right')

# 그래프 출력
plt.show()
```

```python
# 그래프 크기 설정
plt.figure(figsize=(12, 8))

# Model 선택
last_df = result[['Gradient Boosting', 'CNN_LSTM']]

# X축의 위치 설정
x = np.arange(len(last_df))

# 막대그래프 그리기
bar_width = 0.3  # 막대의 너비

plt.bar(x, last_df['Gradient Boosting'], width=bar_width, label='Gradient Boosting', color='limegreen')

plt.bar(x + bar_width, last_df['CNN_LSTM'], width=bar_width, label='CNN_LSTM', color='gold')


plt.xticks(x + 0.1, ['Accuracy', 'Precision', 'Recall', 'F1-Score'])

# X축 레이블 설정
plt.xlabel('Metrics')
plt.ylabel('Scores')
plt.title('Comparison of Model Perfomance')

# 범례 표시
plt.legend(loc='upper right')

# 그래프 출력
plt.show()
```

# **표준편차 시각화**

```python
machine_std = pd.read_csv('machine_learning_result_std_last.csv')
deep_std = pd.read_csv('deep_learning_std_result.csv')
```

```python
result_std = pd.concat([machine_std, deep_std], axis=0).reset_index(drop=True)
a = result_std[4:5]
b = result_std[2:3]
c = result_std[9:10]
d = result_std[10:]

result_std = pd.concat([a, b, c, d], axis=0).reset_index(drop=True)
result_std
```

```python
result_std
```

```python
result_std = result_std.T
result_std.columns = result_std.iloc[0]
result_std = result_std.iloc[1:]
result_std = result_std[:4]
result_std
```

```python
# 그래프 크기 설정
plt.figure(figsize=(12, 8))

# Model 선택
last_df = result_std[['Gradient Boosting', 'Random Forest', 'XGBoost', 'CNN_LSTM', 'CNN', 'CNN_RNN']]

# X축의 위치 설정
x = np.arange(len(last_df))

# 막대그래프 그리기
bar_width = 0.1  # 막대의 너비

plt.bar(x, last_df['Gradient Boosting'], width=bar_width, label='Gradient Boosting', color='green')
plt.bar(x + bar_width, last_df['XGBoost'], width=bar_width, label='XGBoost', color='lime')
plt.bar(x + bar_width * 2, last_df['Random Forest'], width=bar_width, label='Random Forest', color='limegreen')


plt.bar(x + bar_width * 3, last_df['CNN_LSTM'], width=bar_width, label='CNN_LSTM', color='orange')
plt.bar(x + bar_width * 4, last_df['CNN'], width=bar_width, label='CNN', color='gold')
plt.bar(x + bar_width * 5, last_df['CNN_RNN'], width=bar_width, label='CNN_RNN', color='yellow')


plt.xticks(x + 0.3, ['Accuracy', 'Precision', 'Recall', 'F1-Score'])

# X축 레이블 설정
plt.xlabel('Metrics')
plt.ylabel('Scores')
plt.title('Comparison of Model Perfomance')

# 범례 표시
plt.legend(loc='lower right')

# 그래프 출력
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
machine_df=pd.read_csv('machine_learning_result_mean_last.csv')
machine_df
```

```python
deep_df=pd.read_csv('deep_learning_mean_result.csv')
deep_df
```

```python
four_df=machine_df.iloc[[0,2,3,1],:]
four_df
```

```python
df_sorted = four_df.sort_values(by='std', ascending=True)

plt.figure(figsize=(10, 8))

models = df_sorted['Model']
std = df_sorted['std']
colors = ['limegreen', 'forestgreen', 'darkgreen', 'mediumseagreen']

plt.bar(models, std, color=colors, alpha=0.5)

plt.title('Standard Deviation of Model Performance')
plt.xlabel('Model')
plt.ylabel('Standard Deviation')
#plt.xticks(rotation=45, ha='right')  # x축 레이블을 45도로 회전하여 가독성 향상
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
df2_sorted = deep_df.sort_values(by='std', ascending=True)

plt.figure(figsize=(10, 6))

models = df2_sorted['Model']
std = df2_sorted['std']
colors = ['limegreen', 'forestgreen', 'darkgreen', 'mediumseagreen']

plt.bar(models, std, color=colors, alpha=0.5)

plt.title('Standard Deviation of Model Performance')
plt.xlabel('Model')
plt.ylabel('Standard Deviation')
#plt.xticks(rotation=45, ha='right')  # x축 레이블을 45도로 회전하여 가독성 향상
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

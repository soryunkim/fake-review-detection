## **라벨 수를 조정하여 Dataset 생성**
- 모델을 학습하기 위해선 라벨의 개수가 동일한 것이 권장되며, 이에 따라 true data를 샘플링할 예정

```python
# true 데이터를 fake의 개수에 맞게 sampling
true = true.sample(n=len(fake), random_state=42)

print(true.shape)
print(true['fake'].value_counts())
```

```python
true
```

```python
review = pd.concat([true, fake])
review
```
